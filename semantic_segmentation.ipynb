2. What does an encoder do?
--> Early Layers: edges, colors, simple texture;
--> Middle layers: roads, field boundaries, rooftops, vegetation pattern;
--> Deeper layers: Higher level patern level pattern;

Exemple case of how ememcoder work: W assume an input image

Encoder: Input (256 * 256) ---> Original image;

*   Encoder1 128 * 128 -->  Reduide the dimension and Learn basic features;
*   Encoder1 64 * 64 -->  Reduide the dimension and Learn more complex patters;
*   Encoder1 32 * 32 -->  Reduide the dimension and Learn higher level features;
*   Encoder1 16 * 16 --> Show compact and semantic features;










3. What does decoder do ? unbsample features map to increase the spacial resolution  and combine features from the earlier  encoder layers when skip connection are use.
*   Decoder stage:
*   starting  features :---> 16 * 16
*   Decoder1 32 * 32 --> Reduide the dimension and Learn basic features;
*   Decoder2 64 * 64 --> Reduide the dimension and Learn more complex patters;
*   Decoder3 128 * 128 --> Reduide the dimension and Learn higher level features;
*  Final image  256 * 256 --> Show compact and semantic features;
